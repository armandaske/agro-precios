```python
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt

LIBRO_RUTA = Path(r"C:\Users\Dell-G3\Documents\Jupyter-projects\Others\agro-precios\data\analysis\libro_maestro_precios.xlsx")
HOJA_COMPARATIVO = "comparativo_diario_ancho"
HOJA_COBERTURA = "cobertura"
HOJA_COMPARATIVO_ALIASES = [HOJA_COMPARATIVO, "compare_daily_wide"]
HOJA_COBERTURA_ALIASES = [HOJA_COBERTURA, "coverage"]

def leer_hoja_con_aliases(ruta: Path, aliases: list[str]) -> pd.DataFrame:
    with pd.ExcelFile(ruta) as libro:
        for hoja in aliases:
            if hoja in libro.sheet_names:
                return pd.read_excel(ruta, sheet_name=hoja)
    raise ValueError(f"No se encontro ninguna hoja compatible en {ruta}: {aliases}")
```

```python
comparativo_df = leer_hoja_con_aliases(LIBRO_RUTA, HOJA_COMPARATIVO_ALIASES)
cobertura_df = leer_hoja_con_aliases(LIBRO_RUTA, HOJA_COBERTURA_ALIASES)

compare_df["run_date"] = pd.to_datetime(compare_df["run_date"])
compare_df.head()
```
